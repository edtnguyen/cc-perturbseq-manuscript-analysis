# Figure 3 — Hippo signaling is required for efficient endothelial differentiation

**Implemented here:** panel d. Panels b–c remain in their UMAP/neighborhood and guide-level EC workflows.

Canonical source inspected: `programDE/notebooks/program_annotation_panels_D0_D1_D2_D3_rotated_logos_D2_extra_D3_trimmed_markers.ipynb`. The port uses its three aligned data layers: programDE effects, cNMF spectra, and motif enrichment.


```python
from pathlib import Path
import matplotlib.pyplot as plt
import pandas as pd

from src.cc_perturbseq.figure_notebook import (
    get_path, initialize_figure, load_figure_manifest, panel_directory, write_receipt,
)
from src.cc_perturbseq.program_panels import *

def export_figure(fig, stem, panel):
    out = panel_directory(ctx, panel)
    paths = {}
    for suffix, kwargs in [("pdf", {}), ("svg", {}), ("png", {"dpi": 300})]:
        path = out / f"{stem}.{suffix}"
        fig.savefig(path, **kwargs)
        paths[suffix] = str(path)
    return paths

FIGURE = load_figure_manifest("Figure_03_panels_b_d")
ctx = initialize_figure(FIGURE)
ctx["input_audit"]
```

## Panel d — Integrated P2/P34 annotation


```python
effect_path = get_path(ctx["paths"], "programde.effect_matrix_all_days")
q_path = get_path(ctx["paths"], "programde.qvalue_matrix_all_days", required=False)
effects = load_program_effects(effect_path, q_path, filename=config_value(ctx["paths"], "programde.effects_filename", required=False))
spectra_path = get_path(ctx["paths"], "ccperturbseq.cnmf_spectra")
spectra = load_spectra(spectra_path)
long_motif_path = get_path(ctx["paths"], "multiome.enhancer_enrichment_results", required=False)
long_motif_path = long_motif_path if long_motif_path and long_motif_path.exists() else None
motifs = load_motif_enrichment(
    long_motif_path,
    fdr_template=config_value(ctx["paths"], "multiome.motif_fdr_template", required=False),
    score_template=config_value(ctx["paths"], "multiome.motif_score_template", required=False),
)
fig, tables = plot_integrated_annotation(
    effects, spectra, motifs,
    day="D3", programs=[2, 34],
    targets=["NF2", "LATS2", "AMOTL2", "TAOK1"],
    genes=["CDH5", "NRP1", "EFNB2", "TAGLN", "KRT8", "ACTC1", "ANKRD1"],
    motif_names=config_value(ctx["paths"], "figure_options.figure03d.motifs", required=False),
    program_labels={2: "P2 Artery endothelium", 34: "P34 Mesenchyme, YAP/TAZ-induced"},
)
outputs = export_figure(fig, "Figure_03d_program_integration", "d")
source_tables = save_source_tables(tables, panel_directory(ctx, "d"), "Figure_03d")
write_receipt(ctx, "d", status="IMPLEMENTED_NOT_YET_NUMERICALLY_VALIDATED", inputs={"effects": str(effect_path), "spectra": str(spectra_path), "motifs": str(long_motif_path)}, outputs={**outputs, **source_tables}, parameters={"day": "D3", "programs": [2,34]})
plt.show()
```

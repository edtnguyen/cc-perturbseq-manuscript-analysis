# Figure 3 — Hippo signaling is required for efficient endothelial differentiation

Second-pass implementation of panels b–c. Panel d is deferred because the final signed motif matrix, motif ordering and logo selection remain unresolved.


```python
import matplotlib.pyplot as plt
from src.cc_perturbseq.figure_notebook import get_path, initialize_figure, load_figure_manifest, panel_directory, write_receipt
from src.cc_perturbseq.program_panels import *

def export(fig, panel, stem):
    out = panel_directory(ctx, panel)
    result = {}
    for suffix, kwargs in [("pdf", {}), ("svg", {}), ("png", {"dpi": 300})]:
        path = out / f"{stem}.{suffix}"
        fig.savefig(path, **kwargs)
        result[f"{stem}_{suffix}"] = str(path)
    return result

FIGURE = load_figure_manifest("Figure_03_panels_b_d")
ctx = initialize_figure(FIGURE)
ctx["input_audit"]
```

## Panel b — programs, markers and Hippo carrier neighborhoods


```python
adata_path = get_path(ctx["paths"], "ccperturbseq.integrated_anndata")
usage_path = get_path(ctx["paths"], "ccperturbseq.cnmf_usage")
metadata_path = get_path(ctx["paths"], "ccperturbseq.target_metadata", required=False)
adata, usage = load_integrated_anndata(adata_path, usage_path)
features = [("program",2),("program",34),("gene","CDH5"),("gene","ACTC1")]
fig_features, feature_tables = plot_umap_feature_grid(adata, usage, features)
targets = ["NF2", "LATS2", "AMOTL2", "TAOK1"]
guide_map = load_guide_target_map(adata, metadata_path)
neighborhoods = carrier_seeded_neighborhood(adata, usage, guide_map, targets, k=300, pseudocount=0.5)
fig_neighborhoods = plot_carrier_neighborhood_grid(adata, neighborhoods, targets)
outputs = {**export(fig_features, "b", "Figure_03b_program_marker_umaps"), **export(fig_neighborhoods, "b", "Figure_03b_hippo_carrier_neighborhoods")}
tables = save_source_tables({**feature_tables, "neighborhood_statistics": neighborhoods}, panel_directory(ctx, "b"), "Figure_03b")
write_receipt(ctx, "b", status="IMPLEMENTED_NOT_YET_NUMERICALLY_VALIDATED", inputs={"anndata": str(adata_path), "usage": str(usage_path), "guide_metadata": str(metadata_path)}, outputs={**outputs, **tables}, parameters={"features": features, "targets": targets, "k": 300, "same_day": True, "pseudocount": 0.5})
plt.show()
```

## Panel c — guide-level Day 3 EC fractions


```python
ec_path = get_path(ctx["paths"], "ccperturbseq.ec_fraction_table")
ec_table = standardize_ec_fraction_table(ec_path)
targets_c = ["NF2", "LATS2", "AMOTL2", "TAOK1"]
fig, source_c = plot_guide_ec_fraction(ec_table, targets=targets_c)
outputs = export(fig, "c", "Figure_03c_hippo_guide_ec_fraction")
tables = save_source_tables({"guide_ec_fraction": source_c}, panel_directory(ctx, "c"), "Figure_03c")
write_receipt(ctx, "c", status="IMPLEMENTED_NOT_YET_NUMERICALLY_VALIDATED", inputs={"ec_fraction_table": str(ec_path)}, outputs={**outputs, **tables}, parameters={"targets": targets_c, "point_unit": "guide"})
plt.show()
```

# Figure S7 — PKC isoform effects and expression during hiPSC-to-EC differentiation

Second-pass implementation of both panels using the genome-wide screen outputs, raw CC-Perturb-seq counts and GENCODE v43 union-exon lengths.


```python
import matplotlib.pyplot as plt
from src.cc_perturbseq.figure_notebook import get_path, initialize_figure, load_figure_manifest, panel_directory, write_receipt
from src.cc_perturbseq.program_panels import *

def export(fig, panel, stem):
    out = panel_directory(ctx, panel)
    result = {}
    for suffix, kwargs in [("pdf", {}), ("svg", {}), ("png", {"dpi": 300})]:
        path = out / f"{stem}.{suffix}"
        fig.savefig(path, **kwargs)
        result[suffix] = str(path)
    return result

FIGURE = load_figure_manifest("Figure_S07")
ctx = initialize_figure(FIGURE)
ctx["input_audit"]
```

## Panel a — genome-wide-screen effects


```python
mageck_path = get_path(ctx["paths"], "screen.genomewide_mageck_pos_vs_neg")
utest_path = get_path(ctx["paths"], "screen.genomewide_utest_pos_vs_neg")
screen = load_genomewide_screen(mageck_path, utest_path)
fig_a, source_a = plot_pkc_screen(screen)
outputs = export(fig_a, "a", "Figure_S07a_pkc_screen_effects")
tables = save_source_tables({"pkc_screen_effects": source_a}, panel_directory(ctx, "a"), "Figure_S07a")
write_receipt(ctx, "a", status="IMPLEMENTED_NOT_YET_NUMERICALLY_VALIDATED", inputs={"mageck": str(mageck_path), "utest": str(utest_path)}, outputs={**outputs, **tables}, parameters={"canonical_pkc_isoforms": PKC_GROUPS})
plt.show()
```

## Panel b — day-level pseudobulk TPM


```python
adata_path = get_path(ctx["paths"], "ccperturbseq.integrated_anndata")
gtf_path = get_path(ctx["paths"], "reference.gencode_v43_gtf")
adata, _ = load_integrated_anndata(adata_path)
tpm = calculate_pkc_pseudobulk_tpm(adata, gtf_path, count_layer="raw_counts")
fig_b = plot_pkc_tpm(tpm)
outputs = export(fig_b, "b", "Figure_S07b_pkc_pseudobulk_tpm")
tables = save_source_tables({"pkc_tpm": tpm}, panel_directory(ctx, "b"), "Figure_S07b")
write_receipt(ctx, "b", status="IMPLEMENTED_NOT_YET_NUMERICALLY_VALIDATED", inputs={"anndata": str(adata_path), "gtf": str(gtf_path)}, outputs={**outputs, **tables}, parameters={"count_layer": "raw_counts", "gene_length": "union exon length"})
plt.show()
```

# Figure S5 — Multiome quality control, ChromBPNet performance and motif enrichment

Second-pass implementation of panels a–f and h. Panel g remains an editable analysis schematic.


```python
import matplotlib.pyplot as plt
from src.cc_perturbseq.figure_notebook import get_path, initialize_figure, load_figure_manifest, panel_directory, source_asset_status, write_receipt
from src.cc_perturbseq.program_panels import *

def export(fig, panel, stem):
    out = panel_directory(ctx, panel)
    result = {}
    for suffix, kwargs in [("pdf", {}), ("svg", {}), ("png", {"dpi": 300})]:
        path = out / f"{stem}.{suffix}"
        fig.savefig(path, **kwargs)
        result[f"{stem}_{suffix}"] = str(path)
    return result

FIGURE = load_figure_manifest("Figure_S05")
ctx = initialize_figure(FIGURE)
ctx["input_audit"]
```

## Panels a–c — multiome QC


```python
qc_path = get_path(ctx["paths"], "multiome.qc_tables")
tss_path = get_path(ctx["paths"], "multiome.tss_profiles")
qc = load_multiome_qc(qc_path)
fig_ac, source_ac = plot_multiome_qc(qc, tss_path)
outputs = export(fig_ac, "a-c", "Figure_S05a_c_multiome_qc")
tables = save_source_tables(source_ac, panel_directory(ctx, "a-c"), "Figure_S05a_c")
write_receipt(ctx, "a-c", status="IMPLEMENTED_NOT_YET_NUMERICALLY_VALIDATED", inputs={"qc_tables": str(qc_path), "tss_profiles": str(tss_path)}, outputs={**outputs, **tables}, parameters={"tss_flank_normalization": True})
plt.show()
```

## Panels d–e — ChromBPNet held-out performance


```python
performance_path = get_path(ctx["paths"], "multiome.chrombpnet_performance")
performance = standardize_chrombpnet_performance(performance_path)
fig_d, fig_e, source_d, source_e = plot_chrombpnet_performance(performance)
outputs = {**export(fig_d, "d-e", "Figure_S05d_observed_predicted"), **export(fig_e, "d-e", "Figure_S05e_model_performance")}
tables = save_source_tables({"observed_predicted": source_d, "fold_summary": source_e}, panel_directory(ctx, "d-e"), "Figure_S05d_e")
write_receipt(ctx, "d-e", status="IMPLEMENTED_NOT_YET_NUMERICALLY_VALIDATED", inputs={"performance": str(performance_path)}, outputs={**outputs, **tables}, parameters={"cross_validation_folds": 5})
plt.show()
```

## Panel f — motif discovery QC


```python
motif_qc_path = get_path(ctx["paths"], "multiome.motif_qc")
motif_qc = standardize_motif_qc(motif_qc_path)
fig_f = plot_motif_qc(motif_qc)
outputs = export(fig_f, "f", "Figure_S05f_motif_qc")
tables = save_source_tables({"motif_qc": motif_qc}, panel_directory(ctx, "f"), "Figure_S05f")
write_receipt(ctx, "f", status="IMPLEMENTED_NOT_YET_NUMERICALLY_VALIDATED", inputs={"motif_qc": str(motif_qc_path)}, outputs={**outputs, **tables}, parameters={})
plt.show()
```

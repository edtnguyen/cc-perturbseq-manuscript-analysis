# Figure S8 — Distinct regulator groups alter EC differentiation through different program states

Second-pass implementation of panels b–g. Panel a remains the editable BMP-signaling schematic.


```python
import matplotlib.pyplot as plt
from src.cc_perturbseq.figure_notebook import get_path, initialize_figure, load_figure_manifest, panel_directory, source_asset_status, write_receipt
from src.cc_perturbseq.program_panels import *

def export(fig, panel, stem):
    out = panel_directory(ctx, panel)
    result = {}
    for suffix, kwargs in [("pdf", {}), ("svg", {}), ("png", {"dpi": 300})]:
        path = out / f"{stem}.{suffix}"
        fig.savefig(path, **kwargs)
        result[f"{stem}_{suffix}"] = str(path)
    return result

FIGURE = load_figure_manifest("Figure_S08")
ctx = initialize_figure(FIGURE)
ctx["input_audit"]
```

```python
source_asset_status(ctx, "a", "assets.s8a_bmp_schematic")
```

## Panels b and e — guide-level Day 3 EC fractions


```python
ec_path = get_path(ctx["paths"], "ccperturbseq.ec_fraction_table")
ec_table = standardize_ec_fraction_table(ec_path)
bmp_targets = ["ACVR2A", "BMPR1A", "SMAD1", "SMAD5"]
mesoderm_targets = ["EOMES", "MIXL1", "ACVR1B", "SMAD3"]
fig_b, source_b = plot_guide_ec_fraction(ec_table, targets=bmp_targets)
fig_e, source_e = plot_guide_ec_fraction(ec_table, targets=mesoderm_targets)
outputs_b = export(fig_b, "b", "Figure_S08b_bmp_ec_fraction")
tables_b = save_source_tables({"guide_ec_fraction": source_b}, panel_directory(ctx, "b"), "Figure_S08b")
write_receipt(ctx, "b", status="IMPLEMENTED_NOT_YET_NUMERICALLY_VALIDATED", inputs={"ec_fraction_table": str(ec_path)}, outputs={**outputs_b, **tables_b}, parameters={"targets": bmp_targets})
outputs_e = export(fig_e, "e", "Figure_S08e_mesoderm_ec_fraction")
tables_e = save_source_tables({"guide_ec_fraction": source_e}, panel_directory(ctx, "e"), "Figure_S08e")
write_receipt(ctx, "e", status="IMPLEMENTED_NOT_YET_NUMERICALLY_VALIDATED", inputs={"ec_fraction_table": str(ec_path)}, outputs={**outputs_e, **tables_e}, parameters={"targets": mesoderm_targets})
plt.show()
```

## Panels c and f — program and marker UMAPs


```python
adata_path = get_path(ctx["paths"], "ccperturbseq.integrated_anndata")
usage_path = get_path(ctx["paths"], "ccperturbseq.cnmf_usage")
adata, usage = load_integrated_anndata(adata_path, usage_path)
features_c = [("program",2),("program",33),("gene","CDH5"),("gene","EOMES")]
features_f = [("program",17),("program",34),("gene","SLC8A1"),("gene","ACTC1")]
fig_c, source_c = plot_umap_feature_grid(adata, usage, features_c)
fig_f, source_f = plot_umap_feature_grid(adata, usage, features_f)
outputs_c = export(fig_c, "c", "Figure_S08c_bmp_program_marker_umaps")
tables_c = save_source_tables(source_c, panel_directory(ctx, "c"), "Figure_S08c")
write_receipt(ctx, "c", status="IMPLEMENTED_NOT_YET_NUMERICALLY_VALIDATED", inputs={"anndata": str(adata_path), "usage": str(usage_path)}, outputs={**outputs_c, **tables_c}, parameters={"features": features_c})
outputs_f = export(fig_f, "f", "Figure_S08f_mesoderm_program_marker_umaps")
tables_f = save_source_tables(source_f, panel_directory(ctx, "f"), "Figure_S08f")
write_receipt(ctx, "f", status="IMPLEMENTED_NOT_YET_NUMERICALLY_VALIDATED", inputs={"anndata": str(adata_path), "usage": str(usage_path)}, outputs={**outputs_f, **tables_f}, parameters={"features": features_f})
plt.show()
```

# Figure S1 — Genome-wide and focused CRISPRi screens

Panels c–d are implemented from the recovered validation-screen analysis. Panels a–b remain original flow-cytometry assets.


```python
# Repository-path bootstrap for local support modules.
from pathlib import Path
import sys


def _find_repo_root(start=None):
    start = Path(start or Path.cwd()).resolve()
    for candidate in (start, *start.parents):
        if (candidate / "src" / "cc_perturbseq").is_dir():
            return candidate
    raise FileNotFoundError(
        "Could not locate the repository root containing "
        "src/cc_perturbseq. Run this notebook from within the "
        "cc-perturbseq-manuscript-analysis checkout."
    )


REPO_ROOT = _find_repo_root()
for _support_path in (REPO_ROOT, REPO_ROOT / "src"):
    _support_path_text = str(_support_path)
    if _support_path_text not in sys.path:
        sys.path.insert(0, _support_path_text)

print(
    "Repository support paths added:",
    REPO_ROOT,
    "and",
    REPO_ROOT / "src",
)
```

```python
import matplotlib.pyplot as plt
from src.cc_perturbseq.figure_notebook import get_path, initialize_figure, load_figure_manifest, panel_directory, source_asset_status, write_receipt
from src.cc_perturbseq.program_panels import *

def export(fig, panel, stem):
    out = panel_directory(ctx, panel)
    result = {}
    for suffix, kwargs in [("pdf", {}), ("svg", {}), ("png", {"dpi": 300})]:
        path = out / f"{stem}.{suffix}"
        fig.savefig(path, **kwargs)
        result[suffix] = str(path)
    return result

FIGURE = load_figure_manifest("Figure_S01")
ctx = initialize_figure(FIGURE)
ctx["input_audit"]
```

## Panels c–d — validation reproducibility


```python
summary_path = get_path(ctx["paths"], "screen.validation_summary")
types_path = get_path(ctx["paths"], "screen.validation_gene_types", required=False)
validation = standardize_validation_summary(summary_path, types_path)
fig_c, source_c, r_c = plot_validation_replicates(validation)
fig_d, source_d, r_d = plot_genomewide_validation(validation)
outputs_c = export(fig_c, "c", "Figure_S01c_validation_replicates")
tables_c = save_source_tables({"replicate_points": source_c}, panel_directory(ctx, "c"), "Figure_S01c")
write_receipt(ctx, "c", status="IMPLEMENTED_NOT_YET_NUMERICALLY_VALIDATED", inputs={"validation_summary": str(summary_path)}, outputs={**outputs_c, **tables_c}, parameters={"correlation": "Pearson", "r": float(r_c)})
outputs_d = export(fig_d, "d", "Figure_S01d_genomewide_validation")
tables_d = save_source_tables({"cross_screen_points": source_d}, panel_directory(ctx, "d"), "Figure_S01d")
write_receipt(ctx, "d", status="IMPLEMENTED_NOT_YET_NUMERICALLY_VALIDATED", inputs={"validation_summary": str(summary_path), "gene_types": str(types_path)}, outputs={**outputs_d, **tables_d}, parameters={"correlation": "Pearson", "r": float(r_d)})
plt.show()
```

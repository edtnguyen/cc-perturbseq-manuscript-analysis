# Figure S9 — On-target CRISPRi knockdown of Hippo-pathway genes across D0–D3

Second-pass implementation using SCEPTRE fold-change estimates, model-based intervals and day-specific BH q-values.


```python
import matplotlib.pyplot as plt
from src.cc_perturbseq.figure_notebook import get_path, initialize_figure, load_figure_manifest, panel_directory, write_receipt
from src.cc_perturbseq.program_panels import *

def export(fig, panel, stem):
    out = panel_directory(ctx, panel)
    result = {}
    for suffix, kwargs in [("pdf", {}), ("svg", {}), ("png", {"dpi": 300})]:
        path = out / f"{stem}.{suffix}"
        fig.savefig(path, **kwargs)
        result[suffix] = str(path)
    return result

FIGURE = load_figure_manifest("Figure_S09")
ctx = initialize_figure(FIGURE)
ctx["input_audit"]
```

```python
path = get_path(ctx["paths"], "sceptre.on_target_results")
results = standardize_knockdown_results(path)
targets = ["NF2", "LATS2", "AMOTL2", "TAOK1"]
fig, source = plot_knockdown_across_days(results, targets=targets)
outputs = export(fig, "all", "Figure_S09_hippo_knockdown_D0_D3")
tables = save_source_tables({"knockdown": source}, panel_directory(ctx, "all"), "Figure_S09")
write_receipt(ctx, "all", status="IMPLEMENTED_NOT_YET_NUMERICALLY_VALIDATED", inputs={"sceptre_results": str(path)}, outputs={**outputs, **tables}, parameters={"targets": targets, "control_reference": 1.0, "interval": "approximate 95% Wald interval on log2 fold-change scale", "significance": "BH q <= 0.05 within day"})
plt.show()
```

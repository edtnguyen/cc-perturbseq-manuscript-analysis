# CDH5-positive versus CDH5-negative volcano plot

This notebook reproduces the hybrid vector/raster Plotnine figure. Nonsignificant gray points are rasterized, while significant points, labels, leader lines, axes, and legend remain vector. PDF fonts are embedded as TrueType, and SVG text remains `<text>`.

```python
from pathlib import Path
from CDH5pos_vs_CDH5neg_volcano_plotnine import (
    load_input, prepare_data, build_volcano, save_outputs
)

HERE = Path.cwd()
INPUT = HERE / 'source' / 'Pos_vs_Neg.gene_summary.txt'
```

```python
raw = load_input(INPUT)
plot_data, selected = prepare_data(raw)
print(plot_data['significance'].value_counts().sort_index())
selected[['gene', 'log2FC', 'FDR', 'neg_log10_FDR', 'significance']]
```

```python
volcano = build_volcano(plot_data, selected)
volcano
```

```python
saved = save_outputs(volcano, plot_data, selected, HERE)
saved
```

```python
import pandas as pd
import numpy as np
```

```python
df_mageck = pd.read_csv("Pos_vs_Neg.gene_summary.txt", sep="\t")
df_utest = pd.read_csv("combined_utest_P1P2_adj_test.txt", sep="\t")
```

```python
df_mageck
```

```python
# Rename column in df_utest
df_utest = df_utest.rename(columns={'p_value': 'u_test_p_value', 
                                    'log2FC': 'u_test_log2FC', 
                                    'p_value_corrected': 'u_test_p_value_corrected'})
df_utest
```

```python
overlap_count = df_mageck['id'].isin(df_utest['Target']).sum()
print(overlap_count)
```

```python
non_overlap_count = (~df_mageck['id'].isin(df_utest['Target'])).sum()
print(non_overlap_count)
```

```python
non_overlap_count_utest = (~df_utest['Target'].isin(df_mageck['id'])).sum()
print(non_overlap_count_utest)
```

```python
non_overlapping_genes = df_utest.loc[~df_utest['Target'].isin(df_mageck['id']), 'Target']
print(non_overlapping_genes)
```

```python
# This ensures that all genes from both df_mageck and df_utest are retained, even if they don't have matches in the other dataframe. 
merged_df = df_mageck.merge(df_utest, left_on='id', right_on='Target', how='outer')

# Ensure all gene names appear under 'id' by filling NaN values in 'id' with 'Target'
merged_df['id'] = merged_df['id'].fillna(merged_df['Target'])

# Drop the redundant 'Target' column
merged_df = merged_df.drop(columns=['Target'])

merged_df
```

```python
merged_df.to_excel("combined_mageck_u_test_pos_v_neg.xlsx", index=False)
```

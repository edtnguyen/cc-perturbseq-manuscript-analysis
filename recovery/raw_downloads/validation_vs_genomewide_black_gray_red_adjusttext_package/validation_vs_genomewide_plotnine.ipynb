# Validation screen versus genome-wide screen

This notebook reproduces the Plotnine scatter plot using the merged `ec_screen_hits.csv` table. Labels are restricted to the 14 requested genes and are positioned automatically through `geom_text(adjust_text=...)`. Significant-in-both points are black, genome-wide-only points are red, and all remaining points are gray.

```python
from pathlib import Path
import adjustText
from validation_vs_genomewide_plotnine import (
    BASE_NAME, CATEGORY_ORDER, load_input, prepare_data, build_plot, save_outputs
)

HERE = Path.cwd()
INPUT = HERE / 'source' / 'ec_screen_hits.csv'
print('adjustText API:', adjustText.__version__)
```

```python
raw = load_input(INPUT)
plot_data, selected, stats = prepare_data(raw)
print(stats)
print(plot_data['category'].value_counts().reindex(CATEGORY_ORDER))
selected[['gene', 'x', 'y', 'category']]
```

```python
plot = build_plot(plot_data, selected, stats)
plot
```

```python
fig, paths, adjusted_positions = save_outputs(plot, HERE)
adjusted_positions
```

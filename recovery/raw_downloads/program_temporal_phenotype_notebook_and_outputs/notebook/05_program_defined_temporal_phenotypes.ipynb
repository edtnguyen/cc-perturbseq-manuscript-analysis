Important: these analyses are **program-first**. They do not use Leiden cluster identities as the main biological unit. They also avoid relying on ad hoc calibrated difference scores for the main figures. Instead, they use raw curated program effects and binary evidence patterns, such as:

- P17 up **and** P15 down = D2 split/diversion pattern
- P34 up **and** EC endpoint programs down = D3 alternative endpoint diversion
- P48 up = early neuroectoderm-like diversion
- P15 down = vascular progenitor loss
